# Supermart Sales Analysis – Data Analytics with AI

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

df = pd.read_csv('supermart_sales_data.csv', parse_dates=['Order Date','Ship Date'])
df.head()
```

```python
print(df.shape)
print(df.isnull().sum())
print(df.describe(include='all').T)
```

```python
print('Total Sales:', df['Sales'].sum())
print('Total Profit:', df['Profit'].sum())
print('Orders:', df['Order ID'].nunique())
print('Average Order Value:', df['Sales'].mean())
```

```python
df['Order Month']=df['Order Date'].dt.to_period('M').astype(str)
monthly=df.groupby('Order Month').agg(Sales=('Sales','sum'),Profit=('Profit','sum')).reset_index()
monthly
```

```python
category=df.groupby('Category').agg(Sales=('Sales','sum'),Profit=('Profit','sum')).sort_values('Sales',ascending=False)
category
```

```python
plt.figure(figsize=(9,5))
plt.plot(monthly['Order Month'],monthly['Sales'],marker='o')
plt.xticks(rotation=45)
plt.title('Monthly Sales Trend')
plt.tight_layout()
plt.show()
```

```python
monthly['Index']=range(len(monthly))
model=LinearRegression().fit(monthly[['Index']],monthly['Sales'])
future=pd.DataFrame({'Index':range(len(monthly),len(monthly)+3)})
future['Forecast Sales']=model.predict(future[['Index']])
future
```
